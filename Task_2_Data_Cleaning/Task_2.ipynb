Data Cleaning

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
print("Hidden nulls created by conversion:", df['TotalCharges'].isnull().sum())
```

```python
print(df[df['TotalCharges'].isnull()][['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges']])
```

```python
print("All affected customers have tenure 0?",
      (df[df['TotalCharges'].isnull()]['tenure'] == 0).all())
```

```python
hidden_nulls = df['TotalCharges'].isnull().sum()
print(f"⚠️  Hidden nulls created by conversion: {hidden_nulls}")
print(f"✅ All belong to new customers (tenure=0): {(df[df['TotalCharges'].isnull()]['tenure'] == 0).all()}")
print(f"📊 These represent {hidden_nulls / len(df) * 100:.2f}% of the dataset")
```

Replace missing values with 0 (clients with tenure=0 have not been billed)

```python
df['TotalCharges'] = df['TotalCharges'].fillna(0)
```

```python
print(f"Total nulls after filling: {df['TotalCharges'].isnull().sum()}")
print(f"TotalCharges = 0: {(df['TotalCharges'] == 0).sum()} clients")
```
